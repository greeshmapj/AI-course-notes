| Step | Process | Example |

| 7️⃣ | Model / Algorithm | Apply ML model → predict meaning/sentiment |
| 8️⃣ | Output | “Positive” sentiment |

# **3️⃣ Stemming**

Meaning: Reducing words to their root form (may not be grammatically correct).
Why Important: Helps reduce vocabulary size and generalize similar words.

Code (using NLTK PorterStemmer):


```python

from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
words = ["running", "runs", "ran", "easily", "fairly"]
stemmed_words = [stemmer.stem(i) for i in words]
print("Stemmed Words:", stemmed_words)
```

# **4️⃣ Lemmatization**

Meaning: Reduces words to their base or dictionary form (grammatically correct).
Why Important: Produces proper root forms which are more meaningful than stemming.

Code (using NLTK WordNetLemmatizer):

```python
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet #stopwords
import nltk

nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()
words = ["singing", "runs", "easily", "fairly"]
lemmatized = [lemmatizer.lemmatize(w, pos="v") for w in words]
print(lemmatized)
```

Because lemmatize() assumes noun (pos='n') by default.

So:

“runs” → noun form is “run” ✅

“running” → noun form stays “running” ❌ (since it’s not seen as a verb)

✅ Correct way (with POS tagging)

To get the true root form, you should specify the POS tag manually or automatically:

## 🌿 Stemming vs 🍃 Lemmatization

| Feature              | **Stemming**                                                                  | **Lemmatization**                                                                  |
| -------------------- | ----------------------------------------------------------------------------- | ---------------------------------------------------------------------------------- |
| **Definition**       | A process that chops off word endings to get the root form (called a *stem*). | A process that reduces a word to its base or dictionary form (called a *lemma*).   |
| **Method**           | Applies simple **rule-based heuristics** (e.g., remove “ing”, “ed”).          | Uses **vocabulary and morphological analysis** (understands word meaning and POS). |
| **Example**          | *"running" → "run"*<br>*"studies" → "studi"*                                  | *"running" → "run"*<br>*"studies" → "study"*                                       |
| **Output words**     | May not be valid English words (e.g., *“comput”* from *“computing”*).         | Always valid dictionary words.                                                     |
| **Accuracy**         | Faster but less accurate.                                                     | Slower but more accurate.                                                          |
| **Library examples** | `PorterStemmer`, `SnowballStemmer` (in NLTK).                                 | `WordNetLemmatizer` (in NLTK), `spacy` lemmatizer.                                 |
| **Use case**         | When **speed** matters more than accuracy (like quick search).                | When **meaning** and **context** matter (like NLP tasks, translation, etc.).       |

---

### 🧠 In short:
- **Stemming** → Mechanical cutting → “quick and dirty.”  
- **Lemmatization** → Linguistically smart → “slow but meaningful.”

```python
import pandas as pd
from sklearn.model_selection import train_test_split #ml
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
```

```python
data = {
    'review': [
        "I loved this movie,987 it was fantastic!",
        "Absolutely terrible, waste of time.",
        "It was okay, not too bad but not great either.",
        "The performance was brilliant and emotional.",
        "Worst movie ever made.",
        "A masterpiece of storytelling and visuals!",
        "Not good, I didn’t like the plot.",
        "Excellent film! Great direction and acting.",
        "It was boring and too long.",
        "Loved it! Highly recommended.","The movie was  bad, I  dont enjoyed it."
    ],
    'sentiment': ['Positive', 'Negative', 'Neutral', 'Positive', 'Negative',
                  'Positive', 'Negative', 'Positive', 'Negative', 'Positive',"Negative"]
}

df = pd.DataFrame(data)
df
```

```python
import pandas as pd
import nltk
import re #replace
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Download stopwords and lemmatizer resources (only first time)
nltk.download('stopwords')
nltk.download('wordnet')

# Initialize tools
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer() #lemma, dict

# Function to clean text
def preprocess_text(text):
    text = text.lower()                            # 1. Lowercase
    text = re.sub(r'[^a-z\s]', '', text)           # 2. Remove punctuation/numbers (plan d)
    words = text.split()                           # 3. Tokenize
    words = [w for w in words if w not in stop_words]  # 4. Remove stopwords
    words = [lemmatizer.lemmatize(w,pos="v") for w in words]   # 5. Lemmatization
    return ' '.join(words)

# Apply preprocessing
df['clean_review'] = df['review'].apply(preprocess_text)
```

```python
df
```

```python
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(df['clean_review'])
y = df['sentiment']
```

```python
X
```

```python
import pandas as pd
# not nessasary
# Convert sparse matrix to a DataFrame
X_df = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out()
)

# Display the first few rows
X_df.head()
```

```python
X_df.style.background_gradient(cmap='hot')
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```
