# Machine Learning – Definition & Types

## 1. What is Machine Learning?

**Machine Learning (ML)** is a subset of Artificial Intelligence that enables a system to **learn patterns from data** and **make predictions or decisions without being explicitly programmed**.

In ML, the model improves its performance **with experience (data)**.

**Formal Definition (Exam-friendly):**  
Machine Learning is a method of data analysis that automates analytical model building by learning from data.

---

## 2. Types of Machine Learning (High-Level)

Machine Learning is mainly classified into:
1. Supervised Learning
2. Unsupervised Learning
3. Reinforcement Learning
4. semi supervised

*(For most exams & interviews, Supervised Learning is the focus)*

---

## 3. Supervised Learning

Supervised learning uses **labeled data**, where both **input features (X)** and **output/target (y)** are known.

It is divided into:
- Regression
- Classification

---

## 4. Regression

### Definition:
**Regression** is a supervised learning technique used to **predict a continuous numerical value**.

### Key Points:
- Output is **continuous**
- Finds relationship between input and output
- Used for prediction and trend analysis

### Examples:
- House price prediction
- Salary prediction
- Temperature forecasting

### Example Outputs:
- ₹5,00,000
- 27.5°C
- 12.8 kg

---

## 5. Classification

### Definition:
**Classification** is a supervised learning technique used to **predict categorical class labels**.

### Key Points:
- Output is **categorical**
- Assigns data points to predefined classes
- Can be binary or multi-class

### Examples:
- Spam vs Not Spam
- Disease: Yes / No
- Customer churn: Yes / No

### Example Outputs:
- Spam
- Fraud
- Positive / Negative

---

## 6. Regression vs Classification (Quick Comparison)

| Feature | Regression | Classification |
|-------|-----------|---------------|
| Type of Output | Continuous | Categorical |
| Output Examples | Price, Salary | Yes/No, Class |
| Algorithms | Linear Regression, Ridge | Logistic Regression, KNN |
| Evaluation Metrics | MAE, MSE, RMSE, R² | Accuracy, Precision, Recall |

---

## 7. One-Line Exam Notes

- Machine Learning learns patterns from data  
- Regression predicts **continuous values**  
- Classification predicts **class labels**  
- Supervised learning uses **labeled data**

## Regression – Deep Dive with Use Cases
 What is Regression?

Regression is a supervised learning technique used to model the relationship between a dependent variable (target) and one or more independent variables (features). The target is continuous in nature.

##  Types of Regression (Important for Interviews & Exams)

### 1 Simple Linear Regression

- One input feature  
- Straight-line (linear) relationship between input and output  

  

**Use Case:**
- Salary vs Experience  
  (Predicting salary based on years of experience)

---

### 2 Multiple Linear Regression

- More than one input feature  
- Models the combined effect of multiple variables  

  

**Use Case:**
- House price prediction using:
  - Area  
  - Bedrooms  
  - Location (encoded)


image.png

```python
# Step 1: Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Step 2: Create dataset
data = {
    'Area': [500, 600, 800, 1000, 1200],
    'Price': [30000, 35000, 50000, 65000, 80000]
}

df = pd.DataFrame(data)

# Step 3: Split into X and y
X = df[['Area']]      # Independent variable (must be 2D)
y = df['Price']       # Dependent variable

# Step 4: Create and train model
model = LinearRegression()
model.fit(X, y)

# Step 5: Prediction
area_new = [[900]] #feature
predicted_price = model.predict(area_new)

print("Predicted price for 900 sqft:", predicted_price[0])

# Step 6: Visualize
plt.scatter(X, y)
plt.plot(X, model.predict(X))
plt.xlabel("Area (sqft)")
plt.ylabel("Price (Lakhs)")
plt.title("Simple Linear Regression")
plt.show()
```

Polynomial Regression Notes
1️⃣ Definition

Polynomial Regression is a type of regression that models the non-linear relationship between independent variables and the target.

It is an extension of Linear Regression: instead of a straight line, it fits a curve.𝑛
n = degree of the polynomial.

2️⃣ When to Use

Linear regression underfits the data (cannot capture curvature).

Relationship between features and target is non-linear.

Dataset is small/medium (high degree → overfitting risk).

3️⃣ Advantages

Captures non-linear patterns.

Still linear in terms of coefficients (simple to solve).

Works with multiple features.

4️⃣ Disadvantages / Cautions

High degree → overfitting.

Sensitive to outliers.

Feature scaling is recommended for higher-degree polynomials.
