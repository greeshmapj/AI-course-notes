## **What is a Lambda Function?**

A lambda function is a small anonymous function in Python.

It can take any number of arguments but can have only one expression.

It returns the value of the expression automatically

Used for short tasks: Often with map(), filter(), reduce(), or as a quick function argument.

### **lambda arguments: expression**

arguments → Input parameters

expression → Single expression whose value is returned

```python
a=3
b=4
sum=a+b
print(sum)
```

```python
def add(a,b):
  sum=a+b
  print(sum)
add(3,4)
```

# **1. map()**
 – Transform Data
👉 Purpose

Applies a function to each element in an iterable.

```python
numbers = [1, 2, 3, 4]
squares = list(map(lambda x: x**2, numbers))
print(squares)   # Output: [1, 4, 9, 16]
```

```python
names = ["python", "java", "c"]

upper_names = list(map(lambda x: x.upper(), names))
print(upper_names)
```

# **2.filter()** – Select Data
👉 Purpose

Filters elements based on a condition (True/False).

```python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(even_numbers)
```

```python
words = ["cat", "tiger", "lion", "elephant"]

long_words = list(filter(lambda x: len(x) > 4, words))
print(long_words)
```

# **3.reduce()** – Combine Data
👉 Purpose

Reduces an iterable to a single value.

```python
from functools import reduce

numbers = [1, 2, 3, 4]

total = reduce(lambda x, y: x + y, numbers)
print(total)
```

```python
from functools import reduce

numbers = [1, 2, 3, 4]

product = reduce(lambda x, y: x * y, numbers)
print(product)

```

# **practice questions**

1.Write a program to find all duplicate elements in a list.

2.Merge two lists and remove duplicates.

3.Create a list of even numbers from 1 to 50 using list comprehension.


4.Write a function that returns the second largest element in a list.

5.pick vowels from your name

```python
numbers = [1, 2, 3, 2, 4, 5, 1, 6]

duplicates = []
for i in numbers:
    if numbers.count(i) > 1 and i not in duplicates:
        duplicates.append(i)

print("Duplicate elements:", duplicates)
```

```python
#2.Merge two lists and remove duplicates.
lst1 = list(map(int,input("Enter numbers(comma separated)for list1: ").split(',')))
lst2 = list(map(int,input("Enter numbers(comma separated) for list2: ").split(',')))

#Merging lists
lst1.extend(lst2)
print(f"Merged initial lists: {lst1}")

#Duplicate removal using set()
my_list = list(set(lst1))
print(f"list without duplicates: {my_list}")
```

```python
#3.Create a list of even numbers from 1 to 50 using list comprehension.
x = [i for i in range(2,51,2)]
print(f"List of even numbers in the range:{x}")
```

```python
#4.Write a function that returns the second largest element in a list.

def second_largest(lst):
  lst = list(set(lst))   #duplicate removal
  lst.sort()
  return lst[-2]

lst = list(map(int,input("Enter comma separated numbers: ").split(',')))
print("Second largest element:", second_largest(lst))
```

```python
#5.pick vowels from your *name*

name = 'greeshma'
list(name)
vowels = ['a','e','i','o','u']
my_list=[]

for i in name:
  if i in vowels:
    my_list.append(i)
print(f"Vowels in name are:{my_list}")
```
