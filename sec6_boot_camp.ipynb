# **WHILE LOOP – PYTHON NOTES**
### **What is a while loop?**

A while loop is used to repeat a block of code as long as a condition is True


### Flow of while loop

1. Condition is checked

2. If True → loop body executes

3. Condition is checked again

4. If False → loop stops

```python
i = 1
while i <= 5:
    print(i)
    i += 1
```

```python
i = 2
while i <= 20:
    print(i)
    i += 2
```

# break

Why use break?

✔ To stop a loop early

✔ To avoid unnecessary iterations

✔ To exit when a result is found

```python
for i in range(1, 11): #do the same using while loop?
    if i == 6:
        break
    print(i)
```

#### **Stop at first negative number**

Write a program that prints numbers from a list until a negative number appears.

x=[5, 8, 2, -1, 7, 9]

o/p:
5

2


```python
x= [5,8,2,-1,7,9]
for i in x:
  if i>=0:
    print(i)
  else:
    break
```

### Why use continue?

✔ To skip unwanted values
✔ To avoid executing remaining code in current loop cycle
✔ To filter data inside loops

```python
for i in range(1, 11):
    if i == 6:
        continue
    print(i)
```

```python
x=[5, 8, 2, -1, 7, 9]
# print positive numbers only using continue?
# print only positive numbers
x=[5,8,2,-1,7,9]
for i in x:
    if i<0:
        continue
    print(i)
```

### Using continue to filter data: Skip absent students

```python
students = ["Anu", "Absent", "Rahul", "Meera"]

for name in students:
    if name == "Absent":
        continue
    print(name)
```

✅ What is pass?

pass is a null statement in Python.
It does nothing and is used as a placeholder.

🔹 Why use pass?

✔ To avoid syntax errors
✔ To write empty blocks temporarily
✔ When logic will be added later




```python
x=5
if x%2==0:
  pass
else:
  print("odd")
```

What does pass do logically?

👉 Absolutely nothing

• No skipping
• No stopping
• No output
• No change in flow

It just tells Python:

“I know this block is empty. That’s okay.”

```python
marks = [80, 35, 90]

for m in marks:
    if m < 40:
        pass   # later: add fail logic
    else:
        print("won:", m)
```
