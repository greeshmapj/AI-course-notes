## MobileNet and Related Functions – Quick Notes

Here's a summary of the MobileNet model and its related functions in Keras:

### 1. MobileNet
| Feature | Description |
| :--- | :--- |
| **Type** | Pretrained **Convolutional Neural Network (CNN)**. |
| **Training Data** | **ImageNet** dataset (trained on 1,000 different classes, e.g., dog, car, cat). |
| **Key Advantage** | **Lightweight & fast**, designed specifically for **mobile and edge devices**. |
| **Primary Use** | **Image classification** without needing to train a model from scratch (i.e., **Transfer Learning**). |
| **Usage** | `model = MobileNet(weights='imagenet')` |
weights='imagenet' → The model will load weights trained on the ImageNet dataset.

ImageNet has 1,000 classes like dog, cat, car, strawberry, etc.

Using these weights, MobileNet already knows how to recognize these objects.

weights=None → The model will be initialized randomly and not pretrained.

You would need to train it from scratch on your own dataset.
---

### 2. `preprocess_input`
| Feature | Description |
| :--- | :--- |
| **Purpose** | Prepares an image array for input into the MobileNet model. |
| **Transformation** | Converts pixel values from the standard **0–255** range to **-1 to 1**. |
| **Importance** | **Normalization** ensures all input values (R, G, B channels) are on the same scale, which is crucial for the model's accuracy. This is the specific input format MobileNet expects. |
| **Usage** | `img_array = preprocess_input(img_array)` |

---

### 3. `decode_predictions`
| Feature | Description |
| :--- | :--- |
| **Purpose** | Translates the raw output of the model (1000 class scores/probabilities) into **human-readable labels**. |
| **Output** | Returns the **top-N predictions** (where N is specified by the `top` argument). |
| **Output Structure** | A list of tuples, each containing: **Class ID** (e.g., `'n07745940'`), **Class Name** (e.g., `'strawberry'`), and **Probability** (e.g., `0.991`). |
| **Usage** | `decode_predictions(preds, top=3)` |
| **Example Interpretation** | The top prediction `'strawberry'` with a probability of $0.991$ means the model is $\approx 99.1\%$ confident the image contains a strawberry. |

---

### 4. Typical Workflow
1.  **Load Model:** Instantiate `MobileNet` with `weights='imagenet'`.
2.  **Load & Resize Image:** Load the image and resize it to the expected MobileNet input shape (typically **224, 224**).
3.  **Prepare Input:** Convert the image to a **NumPy array** and apply **`preprocess_input`**.
4.  **Predict:** Pass the preprocessed array to the model using `model.predict()`.
5.  **Interpret Output:** Use **`decode_predictions`** to get human-readable class names and probabilities.

axis=0 means add a new dimension at the very front.

New shape:

x.shape → (1, 200, 400, 3)


1 → batch size (number of images)

(200, 400, 3) → original image dimensions

```python
import tensorflow as tf #dl library
from tensorflow.keras.applications.mobilenet import MobileNet, preprocess_input, decode_predictions
from tensorflow.keras.preprocessing import image
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Load pretrained MobileNet model
# MobileNet is trained on ImageNet (1000 classes) and can classify common objects.
# weights='imagenet' loads the pretrained weights.
model = MobileNet(weights='imagenet')  # Pretrained model for object classification

# Step 2: Load and display the image
# image.load_img loads an image and resizes it to the target size (224x224 for MobileNet)
img_path = '/content/house.jpg'  # Replace with your uploaded image path
img = image.load_img(img_path, target_size=(224, 224))  # Resize image to 224x224

# Display the uploaded image using matplotlib
plt.imshow(img)
plt.axis('off')  # Hide axes
plt.title("Uploaded Image")
plt.show()
zsv   r
# Step 3: Preprocess image for MobileNet
# Convert the image to a numpy array
x = image.img_to_array(img)

# Expand dimensions to add a batch dimension (MobileNet expects a batch)
x = np.expand_dims(x, axis=0) #[1,224,224,3] #len, width, color(rgb)
#It’s very important that all images in a batch have the same dimensions
# Preprocess the image: normalize pixel values and adjust channels for MobileNet
x = preprocess_input(x)

# Step 4: Make prediction
# The model predicts the probabilities for 1000 ImageNet classes
preds = model.predict(x)

# Step 5: Decode the predictions
# decode_predictions converts model output into human-readable labels with probabilities
decoded = decode_predictions(preds, top=10)[0]  # Top 3 predictions
print("Predictions:")
for i, (imagenetID, label, prob) in enumerate(decoded):
    print(f" {label}: {prob*100:.2f}%")  # Print label and probability
```
