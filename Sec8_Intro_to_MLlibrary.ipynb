```python
#bar diagram to disply the relationship between Student Name & IQ

x = df["Student Name"]
y = df["IQ"]
import matplotlib.pyplot as plt
plt.bar(x,y, color='r' )
plt.xlabel("Marks")
plt.ylabel("IQ")
plt.xticks(rotation=90)
plt.show()
```

```python
#pie diagram to disply the relationship between Student Name & mark
x= df["Student Name"]
y=df["Mark"]
plt.pie(y, autopct="%.2f%%")
plt.legend(x, loc="upper right")

plt.title("Student Name VS Marks")

plt.show()
```
