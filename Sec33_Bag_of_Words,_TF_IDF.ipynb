# 🧩 **Bag of Words (BoW)**

### 📌 Definition
- Bag of Words is a text representation method in NLP.  
- It **converts text into numerical vectors based on word counts**.  
- Ignores grammar, word order, and meaning, only considers frequency.(count)

---

### ⚙️ How It Works
1. **Tokenization:** Split text into words (tokens).  
   Example: `"I love data science"` → `['I', 'love', 'data', 'science']`
2. **Build Vocabulary:** List all unique words in the corpus.  
   Example vocabulary: `['I', 'love', 'data', 'science']`
3. **Vectorization:** Count occurrences of each word in each document.  

| Document | I | love | data | science |
|----------|---|------|------|---------|
| Doc1: "I love data" | 1 | 1 | 1 | 0 |
| Doc2: "I love science" | 1 | 1 | 0 | 1 |

---

### ✅ Advantages
- Simple, easy to implement.  
- Good for small datasets.  

### ❌ Disadvantages
- Ignores context/meaning.  
- High dimensionality for large vocabularies.  
- Doesn’t handle synonyms or polysemy.

- By default, `CountVectorizer` **converts everything to lowercase**. ✅

Punctuation is removed automatically (e.g., !, .) by the tokenizer.

# 📗 TF-IDF (Term Frequency – Inverse Document Frequency)

### 📌 Definition
- TF-IDF is a **numerical statistic** that reflects how important a word is in a document relative to a corpus.  
- It **reduces the weight of common words** (“the”, “is”) and emphasizes words that are more unique to each document.

---

### ⚙️ How It Works

1. **Term Frequency (TF):** Measures how often a word appears in a document.

$$
TF(t,d) = \frac{\text{Count of word } t \text{ in document } d}{\text{Total words in document } d}
$$

2. **Inverse Document Frequency (IDF):** Measures how unique a word is across all documents.

$$
IDF(t) = \log\left(\frac{\text{Total number of documents}}{\text{Number of documents containing } t}\right)
$$

3. **TF-IDF Score:** Multiply TF and IDF to get the final score.

$$
TF\text{-}IDF(t,d) = TF(t,d) \times IDF(t)
$$

> **Note:** In scikit-learn's `TfidfVectorizer`:
> - Each document vector is **L2-normalized**, so the values are scaled between 0 and 1.  
> - IDF uses **smoothing**:  
> $$IDF(t) = \log\frac{1 + n_\text{docs}}{1 + \text{doc count containing t}} + 1$$  
> - This is why TF-IDF scores in Python may differ from textbook examples.

---

### 🧮 Example

Suppose we have 3 documents:

1. `"I love data science"`  
2. `"I love machine learning"`  
3. `"Data science is amazing"`  

- Word **“data”** appears in Doc1 and Doc3 →  
  $$
  IDF(\text{data}) = \log\left(\frac{3}{2}\right) \approx 0.176
  $$

- TF in Doc1 = 1/4 = 0.25  
- TF-IDF(data, Doc1) ≈ 0.044 (textbook calculation)

- **Python Output using scikit-learn:**
$$
\text{IDF} = \log\left(\frac{1 + N}{1 + df}\right) + 1
$$
| amazing | data | is | learning | love | machine | science |
|---------|------|----|----------|------|---------|---------|
| 0       | 0.577 | 0  | 0        | 0.577 | 0       | 0.577 |
| 0       | 0     | 0  | 0.623    | 0.474 | 0.623   | 0      |
| 0.563   | 0.428 |0.563| 0        | 0    | 0       | 0.428 |

- Rows = documents, Columns = words, Values = TF-IDF scores  
- Higher score → word is **more important** in that document.  
- Example: Doc1 → `"data"`, `"love"`, `"science"` are most important.  
- Example: Doc3 → `"amazing"` has the highest TF-IDF → unique word in that doc.

✅ Words that appear in many documents (like `"I"` or `"love"`) get **lower weights**.


### 🧮 L2 Normalization Example

For each document, divide each TF–IDF value by its **L2 norm**.

$$
\text{Norm} = \sqrt{\sum (\text{TF-IDF})^2}
$$

Then the normalized value is:

$$
\text{Normalized TF-IDF} = \frac{\text{TF-IDF}}{\text{Norm}}
$$

---

### 📄 Document 1: “love data science”

TF–IDF values = [0.322, 0.322, 0.322]

$$
\text{Norm} = \sqrt{(0.322)^2 + (0.322)^2 + (0.322)^2}
$$

$$
\text{Norm} = \sqrt{3 \times (0.322)^2} = \sqrt{3 \times 0.1037} = \sqrt{0.311} = 0.557
$$

---

$$
\text{Normalized TF-IDF} = \frac{0.322}{0.557} = 0.577
$$

✅ **Final Normalized TF–IDF vector:** [0.577, 0.577, 0.577]




# extra notes

##**How scikit-learn gets TF-IDF for "data" in Doc1**

Document:
"I love data science"

---

#### Step 1: Term Frequency (TF)
scikit-learn uses **raw count**

TF(data, Doc1) = 1

---

#### Step 2: Smoothed IDF
Formula used:
IDF = log((1 + N) / (1 + df)) + 1

N = 3 (documents)  
df = 2 ("data" appears in Doc1 & Doc3)

IDF(data) = log(4 / 3) + 1 ≈ 1.287

---

#### Step 3: TF × IDF
TF-IDF(data) = 1 × 1.287 = 1.287

---

#### Step 4: Build vector for Doc1
Words in Doc1:
[data, love, science]

All have same df → same IDF

Vector = [1.287, 1.287, 1.287]

---

#### Step 5: L2 Normalization
Formula:
value = x / sqrt(x₁² + x₂² + x₃²)

Denominator:
sqrt(1.287² + 1.287² + 1.287²)
= sqrt(3 × 1.287²)
= 1.287 × sqrt(3)

---

#### Step 6: Final value
data = 1.287 / (1.287 × √3)
     = 1 / √3
     ≈ 0.577

---

### Final Answer
scikit-learn TF-IDF(data, Doc1) = 0.577

## 🧩 TF-IDF Mini Project: Extract Important Words from Documents

### Project Idea
- **Input:** A small set of text documents (like news articles, dialogues, or student essays).  
- **Goal:** Identify the **top N important words** in each document using TF-IDF.


```python
# Sample documents
docs = [
    "Natural Language Processing enables computers to understand human language.",
    "Machine learning algorithms improve automatically through experience.",
    "Artificial intelligence is the simulation of human intelligence processes by machines."
]
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Create TF-IDF vectorizer
vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform(docs)


```

```python
# Feature names (words)
feature_names = vectorizer.get_feature_names_out()
print("Vocabulary:", feature_names)
```

```python
#[2,4,1]
[2,0,1][-2:][::-1] #output is index
```

```python
# Function to get top N words by TF-IDF for each document
def top_n_words(tfidf_matrix, feature_names, top_n=1):
    top_words = []  # Initialize list to store top words per document

    # Loop over each document's TF-IDF vector
    for row in tfidf_matrix.toarray():
      #[0,.55,0,.55,0,.55] # sort [0,0,0,.55,.56,.57]
      # Convert sparse matrix to dense array
        # Get indices of top N TF-IDF scores in descending order
        top_idx = row.argsort()[-top_n:][::-1]

        # Map indices to actual words using feature_names
        words = [feature_names[i] for i in top_idx]

        # Append the list of top words for this document to the result
        top_words.append(words)

    # Return list of top words for all documents
    return top_words

# Apply the function to get top 3 words per document
top_words_per_doc = top_n_words(tfidf_matrix, feature_names, top_n=5)

# Print top words for each document
for i, words in enumerate(top_words_per_doc):
    print(f"Top words in Document {i+1}: {words}")
```

```python
# 📌 Import necessary libraries
import pickle  # For saving/loading Python objects
from sklearn.feature_extraction.text import TfidfVectorizer  # To convert text into TF-IDF vectors
from nltk.tokenize import sent_tokenize  # To split text into sentences
from sklearn.metrics.pairwise import cosine_similarity  # To compute similarity between vectors
import numpy as np  # For numerical operations

# 📌 Function to summarize text
def summarize_text(text, vectorizer=None, top_n=2):
    # Split the text into sentences
    sentences = sent_tokenize(text)

    # If no vectorizer is provided, create a new one
    if vectorizer is None:
        # TF only (use_idf=False), remove common English stopwords
        vectorizer = TfidfVectorizer(use_idf=False, stop_words='english')
        tfidf_matrix = vectorizer.fit_transform(sentences)  # Fit and transform sentences
    else:
        # Use the provided vectorizer to transform sentences
        tfidf_matrix = vectorizer.transform(sentences)

    # Compute cosine similarity between all sentence vectors
    sim_matrix = cosine_similarity(tfidf_matrix, tfidf_matrix)

    # Score each sentence by summing its similarity with all other sentences
    scores = sim_matrix.sum(axis=1)

    # Get indices of top N sentences (highest scores)
    top_idx = scores.argsort()[-top_n:][::-1]

    # Join top sentences to form the summary
    summary = ' '.join(np.array(sentences)[top_idx])

    return summary, vectorizer

# 📌 Example text for training
text = """Natural Language Processing (NLP) is a field of Artificial Intelligence (AI) that enables machines to understand, interpret,
 and generate human language in a meaningful way. It combines knowledge from linguistics, computer science, and AI to analyze and process large amounts
 of text and speech data efficiently. NLP techniques allow computers to perform tasks such as language translation, sentiment analysis, text summarization,
 and question answering. By understanding the structure and meaning of language, NLP helps machines interact with humans more naturally.
 The field continues to grow rapidly, powering applications like chatbots, virtual assistants, and automated content analysis."""

# Generate summary and get the trained TF-IDF vectorizer
summary, vectorizer = summarize_text(text,top_n=2)
print("Summary:", summary)

# 📌 Save the trained TF-IDF vectorizer for future use
with open('swathiii.pkl', 'wb') as file:
    pickle.dump(vectorizer, file)
```

```python
import pandas as pd
import pickle

# Load saved vectorizer
with open('swathi.pkl', 'rb') as f:
    loaded_vectorizer = pickle.load(f)

# Suppose your new text data is in a DataFrame
import pandas as pd

data = {
    'text': [
        """Person A: Hey, did you finish watching that AI documentary last night?
        Person B: Yeah, it was fascinating! They talked about how machine learning models are trained using tons of data.
        Person A: I loved the part where they explained natural language processing — it’s crazy how machines can now write poems.
        Person B: True! But they also mentioned ethical concerns, especially regarding data privacy and bias.
        Person A: Yeah, that’s something we all should think about as technology evolves.""",

        """Person A: Have you heard about the new Mars rover project by NASA?
        Person B: Oh yes! It’s amazing how they’re using AI to navigate rough terrain.
        Person A: The camera system is advanced, too — it helps identify potential hazards in real-time.
        Person B: Exactly, and I read that it even sends weather data and soil composition results back to Earth.
        Person A: That’s incredible! AI is literally helping us explore new worlds."""
    ]
}

df = pd.DataFrame(data)


# Summarize each row using loaded vectorizer
def summarize_from_df(row):
    summary, _ = summarize_text(row, loaded_vectorizer)
    return summary

df['summary'] = df['text'].apply(summarize_from_df)

print(df)
```
