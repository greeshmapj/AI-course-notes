# Activation Functions in ANN

---

## 1. What is an Activation Function?
- A mathematical function that decides whether a neuron should be activated (fired) or not.  
- Introduces **non-linearity** into the ANN, allowing the network to learn complex patterns (not just straight lines).  
- Without activation functions, an ANN would behave like a **linear regression model**, no matter how many layers it has.  

**Example:**  
- Suppose you want to classify students into Pass or Fail based on **Marks** and **Study Hours**.  
  - **Linear (no activation):** Only a straight line boundary (e.g., if marks > 50, study hour > 3 its an ideal case, which cant be always true).  
  - **With activation (ReLU, Sigmoid):** Can form curved or complex boundaries (e.g., if study hours are low but marks are high → still pass).  

---

## 2. Why are Activation Functions Important?
- Help the network learn **non-linear relationships**.  
- Allow ANN to model **complex real-world problems** like image recognition, NLP, predictions.  
- Control the **flow of information** from one layer to the next.  
- Make **gradient-based optimization (backpropagation)** possible.  

---

## 3. Role in a Training Epoch
**During one epoch (1 full pass of data through the network):**  

**Forward Pass:**  
- Inputs → multiplied by weights + bias → passed through activation function.  
- Activation decides neuron’s output → sent to next layer.  

**Backward Pass (Backpropagation):**  
- Loss is calculated at the output.  
- Gradients are computed using derivatives of activation functions.  
- Gradients update the weights.  

> Without differentiable activation functions, backpropagation would not work.  

---

## 4. How It Works (Step-by-Step)
1. Input features enter the network.  
2. Each neuron computes a weighted sum:  
   $$ z = w \cdot x + b $$  
3. Activation function is applied:  
   $$ a = f(z) $$  
4. Output passes to the next layer.  
5. Final layer activation (Sigmoid/Softmax) gives the prediction.  
6. Loss is calculated → Backpropagation adjusts weights → Repeat next epoch.  

---“How much should this weight change?”-VANISHING GRADIENT

## 5. Essential Activation Functions in ANN

**(a) ReLU (Rectified Linear Unit)**  
- **Equation:** $$ f(x) = \max(0, x) $$  
- **Use:** Default choice for **hidden layers**  
- **Why:** Fast, avoids vanishing gradient  

**(b) Sigmoid**  
- **Equation:** $$ f(x) = \frac{1}{1 + e^{-x}} $$  
- **Use:** **Output layer** in binary classification  
- **Why:** Converts output into **probability (0–1)**  

**(c) Softmax**  
- **Equation:** $$ f(x_i) = \frac{e^{x_i}}{\sum_j e^{x_j}} $$  
- **Use:** **Output layer** in multi-class classification  
- **Why:** Converts outputs into a **probability distribution** (sums to 1)  

---

## 6. Quick Rule (Cheat Sheet)
- **Hidden Layers → ReLU**  
- **Binary Classification Output → Sigmoid**  
- **Multi-class Classification Output → Softmax**  

---

## 7. Example Case: Study Hours High, Marks Low
- **Without activation (linear):**  
  - “More study hours → always higher chance of passing.”  
  - Can’t handle exceptions like students who study a lot but still don’t score well.  

- **With activation functions (non-linearity):**  
  - Hidden neurons can learn rules like:  
    - “If study hours are high but marks are below 30 → Fail”  
    - “If marks are above 80 → Pass regardless of study hours”  
  - Each neuron detects **different non-linear conditions**.  

**Outcome:**  
- ANN learns **curved/complex decision boundaries**, covering exceptions like:  
  - High study hours but poor marks  
  - Low study hours but naturally talented students

# Why Do We Compile a Model in ANN?

- In Keras (and most deep learning frameworks), **compiling a model** is like **setting up the rules for training**.  

- When you compile a model, you specify **three main things**:  

1. **Loss Function** – Defines how the model’s predictions are compared to the true values.  
   - **Example:**  
     - `binary_crossentropy` → for binary classification  
     - `categorical_crossentropy` → for multi-class classification  
     - `mse` (mean squared error) → for regression  

2. **Optimizer** – Defines **how the weights are updated** during training to minimize the loss.  
   - **Example:** `adam`, `sgd`, `rmsprop`  

3. **Metrics** – Defines what **performance metrics** you want to track during training.  
   - **Example:** `accuracy`, `precision`, `recall`  

---

### In Simple Terms:
- **Compiling = Preparing the model for learning.**  
- Without compiling, the model **doesn’t know**:  
  - How to **measure errors** (loss)  
  - How to **update weights** (optimizer)  
  - How to **evaluate performance** (metrics)


# Optimizers in ANN

---

## 1. What is an Optimizer?
- An optimizer is an algorithm that updates the **weights and biases** of a neural network to **minimize the loss function**.  
- It controls **how the network learns** during training.  
- Without an optimizer, the network **cannot improve its predictions**.  

---

## 2. Why are Optimizers Important?
- Helps the network **converge faster** to the minimum of the loss function.  
- Prevents **overshooting or slow learning**.  
- Determines the **accuracy and efficiency** of training.  

---

## 3. How Optimizers Work
1. Compute **gradients of the loss function** with respect to weights (via backpropagation).  
2. Update weights using a **specific update rule**.  



## 4. Types of Optimizers (Essential)

**(a) Gradient Descent (GD)**  
- **Batch GD:** Uses all training data to compute gradients.  
- **Pros:** Stable updates  
- **Cons:** Slow for large datasets  

**(b) Stochastic Gradient Descent (SGD)**  
- Uses **one sample at a time** to compute gradients.  
- **Pros:** Faster, can escape local minima  
- **Cons:** Noisy updates  

**(c) Mini-Batch Gradient Descent**  
- Uses **small batches** of data (e.g., 32 samples).  
- **Pros:** Balance between speed and stability  

**(d) Adaptive Optimizers**  
1. **Adam (Adaptive Moment Estimation)**  
   - Combines momentum + RMSProp.  
   - Popular default choice in modern deep learning.  
   - Adapts learning rate for each parameter.  

2. **RMSProp**  
   - Adjusts learning rate based on recent gradients.  
   - Good for **RNNs** or non-stationary problems.  

3. **Momentum**  
   - Accelerates gradient descent in **relevant directions**.  
   - Helps avoid getting stuck in shallow minima.  

---

## 5. Quick Rule (Cheat Sheet)
- **Default/modern networks → Adam**  
- **Simple experiments → SGD**  
- **Time-series / RNN → RMSProp**  

---

✅ **Summary:**  
- Optimizers **update weights using gradients** in **backpropogation**.  
- **Choice of optimizer** affects speed, stability, and convergence.  
- **Adam** is generally the go-to for most ANN problems today.


# Regularization in Machine Learning

## 1. What is Regularization?

**Definition:**  
Regularization is a technique used to **prevent overfitting** in machine learning models by adding a penalty term to the loss function.  

**Goal:**  
Encourage the model to learn **simpler weights** that generalize better to unseen data.  

---

## 2. Why Regularization?

- Models with too many parameters (like deep neural networks) can **memorize training data**.  
- Regularization ensures the model captures **important patterns** instead of noise.  

---

## 3. Types of Regularization

### (a) L1 Regularization (Lasso)
Penalty term:  

$$
\text{Loss} = \text{Original Loss} + \lambda \sum_{i=1}^{n} |w_i|
$$

**Effect:** Shrinks weights, and can drive some weights to exactly **zero** → feature selection.  

---

### (b) L2 Regularization (Ridge)
Penalty term:  

$$
\text{Loss} = \text{Original Loss} + \lambda \sum_{i=1}^{n} w_i^2
$$

**Effect:** Shrinks weights smoothly but never to zero → keeps all features, but reduces their influence.  

---

### (c) Elastic Net
Penalty term:  

$$
\text{Loss} = \text{Original Loss} + \lambda_1 \sum_{i=1}^{n} |w_i| + \lambda_2 \sum_{i=1}^{n} w_i^2
$$

- Combines **L1 and L2** regularization.  
- Useful when dealing with **high-dimensional data**.  

---


### (d) Dropout (in Neural Networks)
- Randomly "drops out" (sets to zero) some neurons during training.  
- Prevents the network from becoming too dependent on certain neurons → improves generalization.  

---

### (e) Early Stopping
- Stop training when validation error starts increasing, even if training error keeps decreasing.  
- Prevents overfitting by not letting the model learn noise.  

---

### (f) Data Augmentation (Indirect Regularization)
- Expands training data artificially (e.g., rotate/flip images, add noise).  
- Makes model robust and reduces overfitting.  

---

## 4. Hyperparameter: λ (Regularization Strength)

- **Small λ →** Model is almost unregularized → may overfit.  
- **Large λ →** Model is heavily penalized → may underfit.  
- Must be tuned (using **Cross Validation**).  

---

## 5. Visual Intuition

- **Without regularization:** Model learns complex curves (overfit).  
- **With regularization:** Model learns smoother, simpler functions.  

---

✅ **Key takeaway:**  
Regularization is about finding the balance between **bias and variance** to ensure good **generalization**.
